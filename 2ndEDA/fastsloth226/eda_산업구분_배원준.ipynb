```python
# ==========================================
# 0. 라이브러리 로드
# ==========================================
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ==========================================
# 1. 데이터 로드 및 3대 후보 산업군 분류 매핑
# ==========================================
df = pd.read_csv('/content/drive/MyDrive/ESG프로젝트/KCGS_평가등급_2021-2025_Long.csv')
df = df.drop_duplicates(subset=['기업명', '평가년도'])

power_cable = [
    'HD현대일렉트릭', '엘에스일렉트릭', '대한전선', '효성중공업',
    '가온전선', '일진전기', '대원전선', '제룡전기', '광명전기', '선도전기', 'LS머트리얼즈'
]
battery_materials = [
    'LG에너지솔루션', '삼성SDI', '포스코퓨처엠', '에코프로비엠', '엘앤에프',
    '에코프로머티', 'SK아이이테크놀로지', '롯데에너지머티리얼즈', '더블유씨피', '세방전지',
    '대주전자재료', '솔루스첨단소재'
]
semicon_pcb = [
    '삼성전자', 'SK하이닉스', 'DB하이텍', 'LX세미콘', '리노공업',
    '하나마이크론', '하나머티리얼즈', '해성디에스', '심텍', '대덕전자',
    '이수페타시스', '코리아써키트', '에스앤에스텍', '에프에스티', '네패스',
    'SFA반도체', '시그네틱스', '가온칩스', '칩스앤미디어', '제주반도체'
]

def assign_industry(name):
    if name in power_cable:
        return '전력인프라'
    elif name in battery_materials:
        return '2차전지/소재'
    elif name in semicon_pcb:
        return '반도체/기판'
    return '기타'

df['세부산업군'] = df['기업명'].apply(assign_industry)
target_df = df[df['세부산업군'] != '기타'].copy()

# 등급 수치화 (A+:6 ~ D:1)
grade_map = {'A+': 6, 'A': 5, 'B+': 4, 'B': 3, 'C': 2, 'D': 1}
for col in ['ESG등급', '환경', '사회', '지배구조']:
    if col in target_df.columns:
        target_df[f'{col}_점수'] = target_df[col].map(grade_map)

df_2025 = target_df[target_df['평가년도'] == 2025]

# =========================================================
# Step 1. 2025년 산업군별 통합 ESG 등급 분포 (Grouped Bar)
# =========================================================
fig1 = px.histogram(
    df_2025,
    x='세부산업군',
    color='ESG등급',
    barmode='group',
    category_orders={
        '세부산업군': ['전력인프라', '반도체/기판', '2차전지/소재'],
        'ESG등급': ['A+', 'A', 'B+', 'B', 'C', 'D']
    },
    title='Step 1. 2025년 산업군별 ESG 등급 분포 비교'
)
fig1.update_layout(xaxis_title='세부 산업군', yaxis_title='기업 수', template='plotly_white')
fig1.show()

# =========================================================
# Step 2. E · S · G 세부 영역별 평균 점수 비교 (Heatmap)
# =========================================================
score_cols = [c for c in ['환경_점수', '사회_점수', '지배구조_점수', 'ESG등급_점수'] if c in df_2025.columns]
esg_means = df_2025.groupby('세부산업군')[score_cols].mean().round(2)

fig2 = px.imshow(
    esg_means,
    text_auto=True,
    color_continuous_scale='YlGnBu',
    title='Step 2. 2025년 산업군별 E / S / G 영역 평균 점수 (6점 만점)'
)
fig2.update_layout(xaxis_title='ESG 세부 영역', yaxis_title='세부 산업군', template='plotly_white')
fig2.show()

# =========================================================
# Step 3. 세부 산업군별 E/S/G 점수 분포 비교 (Boxplot)
# =========================================================
fig3 = px.box(
    df_2025,
    x='세부산업군',
    y='ESG등급_점수',
    color='세부산업군',
    points='all',
    hover_data=['기업명', 'ESG등급'],
    title='Step 3. 2025년 세부 산업군별 ESG 종합 점수 분포'
)
fig3.update_layout(yaxis_title='ESG 등급 점수 (1: D ~ 6: A+)', template='plotly_white')
fig3.show()

# =========================================================
# Step 4. 시계열 전이 분석 (2022 -> 2025 등급 변동 추적)
# =========================================================
if 2022 in target_df['평가년도'].values and 2025 in target_df['평가년도'].values:
    piv_trans = target_df.pivot(index=['세부산업군', '기업명'], columns='평가년도', values='ESG등급_점수').dropna(subset=[2022, 2025]).reset_index()
    piv_trans['등급_변화량'] = piv_trans[2025] - piv_trans[2022]

    fig4 = px.bar(
        piv_trans,
        x='기업명',
        y='등급_변화량',
        color='세부산업군',
        title='Step 4. 2022년 대비 2025년 ESG 등급 상승폭 (도약 기업 식별)'
    )
    fig4.add_hline(y=0, line_dash='dash', line_color='red')
    fig4.update_layout(xaxis_tickangle=-45, yaxis_title='등급 단계 변화 (+: 상승, -: 하락)', template='plotly_white')
    fig4.show()
```

낮은 산업군 평균(2.91, C~B 수준)과 지배구조 취약성

전력인프라는 환경(3.55) 대비 사회(3.36)와 지배구조(2.64) 점수가 낮아 전반적인 ESG 성숙도가 아직 3점 미만에 머물러 있습니다.

이는 산업 전체적으로 가이드라인과 밸류업이 절실한 상태임을 명확히 보여줍니다.

극단적인 양극화 (선도 vs 추종의 격차)

산업군 평균은 2.91로 낮지만, Step 1 & Step 4를 보면 엘에스일렉트릭, 대한전선 같은 확실한 선도기업(A~A+)이 존재합니다.

반면 다수의 중소·중견 전력기기 기업들은 C~D 등급에 머물러 있어, "소수의 선도기업 사례를 분석해 다수의 하위 추종기업을 끌어올린다"는 프로젝트 목표와 완벽히 맞아떨어집니다.

효성중공업(-2단계) vs 광명전기(+1단계) 변동성

대형사인 효성중공업의 급격한 등급 하락과 광명전기의 반등 사례를 대조 분석하면, 감점 방지 및 실질적인 등급 상향을 위한 핵심 액션 아이템(피처)을 도출하기에 가장 적합한 산업군입니다.

```python
# ==============================================================================
# [통합] 전력망 · 반도체 · 2차전지 개별 기업별 ESG 종합 + E/S/G 추이 그리드
# ==============================================================================

# 색상 매핑 설정 (종합등급은 눈에 띄는 빨간색)
color_map = {
    '통합 ESG등급': '#e41a1c',  # 빨간색
    'E (환경)': '#2ca02c',      # 초록색
    'S (사회)': '#ff7f0e',      # 주황색
    'G (지배구조)': '#1f77b4'   # 파란색
}

def plot_esg_grid_with_total(df, industry_name, title, cols_wrap=4, chart_height=800):
    # 1. 해당 산업군 필터링
    sub_df = df[df['세부산업군'] == industry_name].copy()

    # 2. 통합등급 + E/S/G 점수 컬럼을 Long-form으로 변환
    melted_df = pd.melt(
        sub_df,
        id_vars=['기업명', '평가년도'],
        value_vars=['ESG등급_점수', '환경_점수', '사회_점수', '지배구조_점수'],
        var_name='지표구분',
        value_name='점수'
    )

    # 범례 및 라벨 매핑
    melted_df['지표구분'] = melted_df['지표구분'].replace({
        'ESG등급_점수': '통합 ESG등급',
        '환경_점수': 'E (환경)',
        '사회_점수': 'S (사회)',
        '지배구조_점수': 'G (지배구조)'
    })

    # 겹침 방지용 미세 지터링 (통합등급은 원본 위치 유지, E/S/G만 미세 분산)
    np.random.seed(42)
    melted_df['점수_jitter'] = melted_df.apply(
        lambda row: row['점수'] if row['지표구분'] == '통합 ESG등급'
        else row['점수'] + np.random.uniform(-0.06, 0.06), axis=1
    )

    # 3. Plotly 그리드 생성
    fig = px.line(
        melted_df,
        x='평가년도',
        y='점수_jitter',
        color='지표구분',
        facet_col='기업명',
        facet_col_wrap=cols_wrap,
        markers=True,
        color_discrete_map=color_map,
        category_orders={'지표구분': ['통합 ESG등급', 'E (환경)', 'S (사회)', 'G (지배구조)']},
        hover_data={'점수_jitter': False, '점수': True, '평가년도': True, '지표구분': True},
        title=title
    )

    # 통합 ESG등급 라인 두께 강조
    for trace in fig.data:
        if trace.name == '통합 ESG등급':
            trace.line.width = 3.5
            trace.marker.size = 8
        else:
            trace.line.width = 1.8
            trace.marker.size = 5

    # Y축 및 레이아웃 설정
    fig.update_yaxes(
        tickvals=[1, 2, 3, 4, 5, 6],
        ticktext=['D', 'C', 'B', 'B+', 'A', 'A+'],
        range=[0.5, 6.5],
        title_text='등급'
    )
    fig.update_xaxes(dtick=1)
    fig.update_layout(
        template='plotly_white',
        height=chart_height,
        legend=dict(
            title='ESG 평가 지표',
            orientation='h',
            yanchor='bottom',
            y=1.02,
            xanchor='right',
            x=1
        )
    )
    return fig

# ------------------------------------------------------------------------------
# 1. 전력인프라 (11개사)
# ------------------------------------------------------------------------------
fig_power = plot_esg_grid_with_total(
    target_df,
    industry_name='전력인프라',
    title='[1. 전력인프라 11개사] 기업별 통합 ESG등급 및 E · S · G 세부 영역 추이 (2021~2025)',
    cols_wrap=4,
    chart_height=800
)
fig_power.show()

# ------------------------------------------------------------------------------
# 2. 반도체/기판 (20개사)
# ------------------------------------------------------------------------------
fig_semi = plot_esg_grid_with_total(
    target_df,
    industry_name='반도체/기판',
    title='[2. 반도체/기판 20개사] 기업별 통합 ESG등급 및 E · S · G 세부 영역 추이 (2021~2025)',
    cols_wrap=5,
    chart_height=1000
)
fig_semi.show()

# ------------------------------------------------------------------------------
# 3. 2차전지/소재 (12개사)
# ------------------------------------------------------------------------------
fig_battery = plot_esg_grid_with_total(
    target_df,
    industry_name='2차전지/소재',
    title='[3. 2차전지/소재 12개사] 기업별 통합 ESG등급 및 E · S · G 세부 영역 추이 (2021~2025)',
    cols_wrap=4,
    chart_height=800
)
fig_battery.show()
```

```python
# ==============================================================================
# 2025년 B등급 미만(C, D) 17개 기업 시가총액 수집 (네이버 금융 스크래핑)
# ==============================================================================
import requests
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup

# 1. 17개 대상 기업 및 고유 종목코드 딕셔너리
code_dict = {
    # 전력인프라 (5개)
    '제룡전기': '033100', '일진전기': '103590', '선도전기': '007610',
    '대원전선': '006340', '광명전기': '017040',

    # 반도체/기판 (11개)
    '하나마이크론': '067310', '코리아써키트': '007810', '칩스앤미디어': '094360',
    '네패스': '033640', 'SFA반도체': '036540', '에프에스티': '036810',
    '에스앤에스텍': '101490', '시그네틱스': '033170', '제주반도체': '080220',
    '가온칩스': '399720', '리노공업': '058470',

    # 2차전지/소재 (1개)
    '더블유씨피': '393890'
}

# 2. 2025년 C, D등급 타깃 추출
c_d_target = target_df[(target_df['평가년도'] == 2025) & (target_df['ESG등급'].isin(['C', 'D']))][
    ['세부산업군', '기업명', 'ESG등급']
].rename(columns={'ESG등급': '2025_ESG등급'}).drop_duplicates()

c_d_target['종목코드'] = c_d_target['기업명'].map(code_dict)

# 3. 네이버 증권 개요 페이지에서 시가총액(억원) 크롤링
def fetch_naver_market_cap(code):
    if pd.isna(code):
        return np.nan
    url = f"https://finance.naver.com/item/main.naver?code={code}"
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }
    try:
        res = requests.get(url, headers=headers, timeout=5)
        soup = BeautifulSoup(res.text, 'html.parser')

        # 시가총액 태그 추출 (id: _market_sum)
        cap_tag = soup.select_one('#_market_sum')
        if cap_tag:
            # 텍스트 내 쉼표, 줄바꿈, 탭 제거 후 float 변환 (단위: 억원)
            raw_text = cap_tag.text.strip().replace(',', '').replace('\t', '').replace('\n', '')
            # '조' 단위 표기가 포함된 경우 처리 (예: 1조 2345억 -> 12345)
            if '조' in raw_text:
                parts = raw_text.split('조')
                trillion = float(parts[0].strip()) * 10000
                billion = float(parts[1].strip()) if parts[1].strip() else 0
                return trillion + billion
            return float(raw_text)
    except Exception:
        pass
    return np.nan

# 4. 시가총액 적용 및 정렬
c_d_target['시가총액_억원'] = c_d_target['종목코드'].apply(fetch_naver_market_cap)
result_df = c_d_target.sort_values(by=['세부산업군', '시가총액_억원'], ascending=[True, False]).reset_index(drop=True)
result_df
```
