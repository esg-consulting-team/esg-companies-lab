```python
import pandas as pd

# 1. 2025년 전기·전자 101개 기업 데이터 필터링
df_elec_2025 = df[(df['평가년도'] == 2025) & (df['업종명'] == '전기·전자')].copy()

# 2. 한국표준산업분류(KSIC) C26 vs C28 분류 함수 정의
def assign_ksic(company_name):
    company_name = str(company_name).strip()

    # C28: 전기장비 제조업 (배터리/이차전지, 전선, 발전기, 조명, 가전제품, 전력기기 등)
    c28_keywords = [
        'LG에너지솔루션', '삼성SDI', 'SK아이이테크놀로지', '엘앤에프', '포스코퓨처엠', '에코프로비엠', 'DN오토모티브', 'LS머트리얼즈', '모베이스전자', '현대모비스', 'HL만도',
        'HD현대일렉트릭', 'LS일렉트릭', 'LS전선', '대한전선', '가온전선', '일진전기', '효성중공업', '제룡전기', '광명전기', '보성파워텍', '서전기전', 'HD현대에너지솔루션', 'KH 필룩스',
        'LG전자', '위니아', '쿠쿠홈시스', '쿠쿠홀딩스', '신일전자', '파セ', '오텍', '위닉스', 'PN풍년', '두산퓨얼셀'
    ]

    # C28 기업 리스트에 해당하는 키워드가 기업명에 포함되어 있으면 C28로 지정
    if any(k in company_name for k in c28_keywords):
        return 'C28: 전기장비 제조업'
    else:
        return 'C26: 전자부품, 컴퓨터, 영상, 음향 및 통신장비 제조업'

# 3. 분류 적용
df_elec_2025['KSIC_분류'] = df_elec_2025['기업명'].apply(assign_ksic)

# 4. 분류별 기업 수 집계 및 출력
ksic_summary = df_elec_2025.groupby('KSIC_분류')['기업명'].agg(
    기업수='nunique',
    기업목록=lambda x: ", ".join(x)
).reset_index()

print("=" * 80)
print("■ 1. 한국표준산업분류(C26 / C28) 대분류별 기업 수 집계")
print("=" * 80)
print(ksic_summary[['KSIC_분류', '기업수']].to_string(index=False))
print(f"\n총 포함 기업 수: {ksic_summary['기업수'].sum()}개사")

print("\n" + "=" * 80)
print("■ 2. 세부 분류별 포함 기업 목록")
print("=" * 80)
for _, row in ksic_summary.iterrows():
    print(f"\n▶ [{row['KSIC_분류']}] - 총 {row['기업수']}개사")
    print(f"   기업 목록: {row['기업목록']}")
```

```python
import pandas as pd

# ----------------------------------------------------
# 1. 대상 기업(C26 업종 76개사) 필터링
# ----------------------------------------------------
c26_companies = [
    '에스앤에스텍', 'LX세미콘', '주성코퍼레이션', '하나머티리얼즈', '서진시스템', '유티아이', '인텔리안테크', '드림텍',
    '해성디에스', '제이앤티씨', '덕산네오룩스', 'RFHIC', '심텍', '솔루엠', '한화시스템', '이녹스첨단소재',
    '솔루스첨단소재', '대덕전자', '더블유씨피', '가온칩스', '에코프로머티', 'SK하이닉스', 'DB하이텍', '금호전기',
    '삼화콘덴서공업', '세방전지', '한솔테크닉스', '써니전자', '씨아이테크', '삼영전자공업', '휴니드테크놀러지스', '삼성전자',
    '대원전선', '선도전기', '이수페타시스', '코리아써키트', '대동전자', '아남전자', '경인전자', '삼성전기',
    '경동나비엔', '삼화전기', '엘에스일렉트릭', 'LG이노텍', '삼화전자공업', '와이투솔루션', '신성이엔지', '성문전자',
    '에이엔피', '광전자', '티에이치엔', '엑시큐어하이트론', '롯데에너지머티리얼즈', '일진디스플', '시노펙스', '한국단자공업',
    '신도리코', '케이엠더블유', '시그네틱스', '자화전자', '네패스', 'LG디스플레이', 'SFA반도체', '에프에스티',
    '주연테크', '서울반도체', '인탑스', '인터플렉스', '리노공업', '에스피지', '신성델타테크', '하나마이크론',
    '다원시스', '대주전자재료', '제주반도체', '아이티엠반도체', '비에이치', '이엠텍', '파트론', 'KEC', '칩스앤미디어', '엠씨넥스'
]

df_c26 = df[(df['기업명'].isin(c26_companies)) & (df['평가년도'].between(2022, 2025))].copy()

# 2025년 신규 편입 6개사 제외 모수 76개사
pv_check = pd.pivot_table(df_c26, index='기업명', columns='평가년도', values='ESG등급', aggfunc='count').fillna(0)
only_2025 = pv_check[(pv_check[2022]==0) & (pv_check[2023]==0) & (pv_check[2024]==0) & (pv_check[2025]==1)].index.tolist()
target_76 = [c for c in c26_companies if c not in only_2025]

df_target = df_c26[df_c26['기업명'].isin(target_76)].copy()

# ----------------------------------------------------
# 2. 등급 수치화 매핑 (변화량 계산용)
# ----------------------------------------------------
grade_map = {'S': 7, 'A+': 6, 'A': 5, 'B+': 4, 'B': 3, 'C': 2, 'D': 1}
cols_to_map = ['ESG등급', '환경', '사회', '지배구조']

for col in cols_to_map:
    if col in df_target.columns:
        df_target[f'{col}_num'] = df_target[col].map(grade_map).fillna(0)

# ----------------------------------------------------
# 3. 턴어라운드 유형별 감지 함수
# ----------------------------------------------------

# (1) 보고서 공시 전환 턴어라운드 (N -> Y 전환)
pv_sus = df_target.pivot_table(index='기업명', columns='평가년도', values='지속가능경영보고서_발행', aggfunc='first').fillna('N')
pv_cg = df_target.pivot_table(index='기업명', columns='평가년도', values='기업지배구조보고서_발행', aggfunc='first').fillna('N')

# 과거(2022~2024)엔 둘 다 N이었다가 2025년에 하나 이상 Y로 바뀐 기업
first_report_comps = []
for comp in target_76:
    past_sus = (pv_sus.loc[comp, [2022, 2023, 2024]] == 'Y').any() if comp in pv_sus.index else False
    past_cg = (pv_cg.loc[comp, [2022, 2023, 2024]] == 'Y').any() if comp in pv_cg.index else False

    curr_sus = pv_sus.loc[comp, 2025] == 'Y' if comp in pv_sus.index else False
    curr_cg = pv_cg.loc[comp, 2025] == 'Y' if comp in pv_cg.index else False

    # 보고서를 공시하기 시작한 기업
    if not (past_sus or past_cg) and (curr_sus or curr_cg):
        first_report_comps.append(comp)

# (2) 종합 ESG 등급 2단계 이상 상승 기업 (예: C -> B+ 이상)
pv_esg_num = df_target.pivot_table(index='기업명', columns='평가년도', values='ESG등급_num', aggfunc='first').fillna(0)
pv_esg_str = df_target.pivot_table(index='기업명', columns='평가년도', values='ESG등급', aggfunc='first').fillna('-')

turnaround_esg = []
for comp in pv_esg_num.index:
    start_score = pv_esg_num.loc[comp, 2022] if 2022 in pv_esg_num.columns else 0
    end_score = pv_esg_num.loc[comp, 2025] if 2025 in pv_esg_num.columns else 0

    # 최초 등급 존재 + 2단계 이상 상승
    if start_score > 0 and (end_score - start_score) >= 2:
        diff = int(end_score - start_score)
        turnaround_esg.append({
            '기업명': comp,
            '2022년': pv_esg_str.loc[comp, 2022],
            '2025년': pv_esg_str.loc[comp, 2025],
            '상승단계': f"+{diff}단계"
        })

df_turnaround_esg = pd.DataFrame(turnaround_esg)

# ----------------------------------------------------
# 4. 결과 분석 출력
# ----------------------------------------------------
print("=" * 75)
print("■ C26 업종 컨설팅용 턴어라운드(Turnaround) 모범사례 기업 추출 결과")
print("=" * 75)

print(f"\n1. [보고서 공시 전환 턴어라운드] (2022~2024 미발행 -> 2025년 첫 공시)")
print(f"   • 대상 기업 ({len(first_report_comps)}개사): {', '.join(first_report_comps) if first_report_comps else '해당 없음'}")
if first_report_comps:
    print("   • 추천 활용: '보고서 공시 시작만으로 등급 상승 기틀을 마련한 대표 사례'")

print(f"\n2. [ESG 종합등급 급상승 턴어라운드] (2022년 대비 2025년 +2단계 이상 상승)")
if not df_turnaround_esg.empty:
    print(df_turnaround_esg.to_string(index=False))
    print("   • 추천 활용: 'C/D 등급에서 2~3년 만에 B+/A 등급으로 진입한 실질 개선 벤치마킹 사례'")
else:
    print("   • 2단계 이상 상승한 기업이 없습니다.")

print("\n" + "=" * 75)
```

우선 분석의 목적: A/A+ 기업 분석은 "이들처럼 되라"고 보여주는 목적이 아니라, ESG 평가항목(E/S/G 각 지표)에서 감점을 받지 않기 위해 갖춰야 할 '필수 요건(Baseline Checklist)'이 무엇인지 파악하기 위해 필요합니다.

### 1. [관점 1] C/D 취약군의 최대 아킬레스건은 ‘지배구조(G)’와 ‘환경(E)’

* **데이터 인사이트**: 레이더 차트에서 C/D 취약군은 사회(S) 영역에 비해 **환경(E)과 지배구조(G) 영역의 면적이 현저히 축소**되어 나타납니다.
* **컨설팅 시사점**:
* 중소·중견기업 다수가 노동/안전/사회공헌 등 **사회(S) 영역**에서는 기본적 법규 준수로 일정 점수를 받지만, **환경(E, 온실가스·폐기물 데이터)** 및 **지배구조(G, 이사회 투명성·주주권익)** 영역에서는 공시 및 제도 부재로 대거 감점을 당하고 있습니다.
* 따라서 C/D 기업 대상 컨설팅 시 "E와 G 항목의 공시 체계 구축"을 최우선 개선 과제(Top Priority)로 지정해야 합니다.



---

### 2. [관점 2] 보고서 발행(Y)은 '평가 등급을 한 단계 격상'시키는 가장 확실한 레버리지

* **데이터 인사이트**: 보고서를 1개 이상 발행한 기업군(Y)의 종합 ESG 및 영역별 평균 점수는 미발행 기업군(N) 대비 **약 1.5 ~ 2.0점(등급 기준 C/D → B/B+ 이상) 이상 압도적으로 높습니다.**
* **컨설팅 시사점**:
* C/D 기업의 낮은 등급 원인은 실제 ESG 경영 수행 여부보다 "공시 부재로 인한 평가 제외(Information Gap)"에 더 큽니다.
* 경영진을 설득할 때 "보고서 발간 하나만으로도 즉시 C/D 탈출 및 B+ 진입이 가능하다"는 명확한 ROI(투자 대비 효과) 근거로 제시할 수 있습니다.

### 3. 영역별 진입 장벽 구조 (G > E > S)

* **데이터 인사이트**: 미발행 기업(N)이 발행 기업(Y)으로 전환할 때 가장 격차가 크게 벌어지는 지표는 지배구조(G)이며, 그다음이 환경(E)입니다.
* **컨설팅 시사점**:
* **지배구조(G)**: '기업지배구조보고서'의 핵심 지표(사외이사 비율, 세부 위원회 설치, 주총 소집 통지 등) 표준화 및 공시 절차 수립이 시급합니다.
* **환경(E)**: Scope 1, 2 온실가스 배출량 산정 및 에너지 정량 데이터 정비 체계(지속가능경영보고서 인덱스 대응)를 조기 확보해야 합니다.
